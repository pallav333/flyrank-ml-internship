### Prioritized Action Queue

| Priority | Recommended Action             | Reason Code                            | Expected Benefit                          |
|----------|--------------------------------|----------------------------------------|-------------------------------------------|
| 1        | Refresh declining content      | Observed decrease in traffic/visibility| Likely recovery of organic presence       |

| 5        | Monitor stable pages           | Consistently high performance observed | Decision support for maintaining current strategy |

### Explanation of Ranking

The actions are ranked to prioritize those with the most immediate and directional impact on performance, based on observed data. 'Refreshing declining content' is ranked highest as it addresses pages experiencing a measured decrease in performance, offering a direct opportunity for recovery. Improving titles and meta descriptions follows, targeting an observed low CTR, which could lead to an increase in user engagement. Adding internal links is next, as it supports content discoverability and is a foundational SEO practice. Updating outdated information addresses content quality and relevance. Finally, 'monitoring stable pages' is crucial for decision support, ensuring continued success and identifying any subtle shifts that might necessitate future action.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

### Who Uses This Playbook

This playbook is intended for SEO specialists, content managers, and marketing teams. It provides decision support for allocating resources to content optimization efforts.

### What Decisions It Supports

This framework supports decisions related to content strategy, including which content to prioritize for updates, how to improve on-page elements, and where to invest in content enhancements. It helps guide editorial calendars and resource allocation based on observed content performance metrics.

### Where It Is Useful
This playbook is useful for identifying content optimization opportunities within a specific sample content-performance dataset. It offers directional insights into content types and attributes that are likely to benefit from specific interventions.

### Limitations

The recommendations provided are based on a sample dataset and historical observations; therefore, they are directional and should not be interpreted as guarantees of specific outcomes, such as Google rankings. The model aims to provide decision support and insights into potential content improvements. It is crucial to understand that these recommendations should support human decision-making rather than being used for automatic publishing or content alteration without expert review. External factors, such as Google algorithm updates or competitor actions, are not explicitly modeled and can influence actual performance. This playbook is a tool for guiding strategy, not a substitute for human expertise.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human Review Checklist

Before taking any recommended action, an SEO or content expert should verify the following:

*   **Content Accuracy:** Ensure the information is still factually correct and aligns with current best practices.
*   **Brand Voice and Tone:** Confirm any proposed changes maintain the established brand voice and tone.
*   **User Intent Alignment:** Verify that the content still effectively addresses the observed user intent.
*   **Competitive Landscape:** Review competitor content for similar topics to identify new opportunities or potential gaps.
*   **Technical SEO Considerations:** Check for any technical SEO issues that might override content optimizations.
*   **Business Objectives:** Ensure the recommended action aligns with current business goals and priorities.

### No-Go List: What Should Never Be Automated

Certain actions carry significant risk and must always remain under human control, never to be automated:

*   **Deleting Pages:** The removal of content or pages should always involve human review to prevent loss of valuable assets or traffic.
*   **Publishing Content:** Automated publishing of new content is not recommended as it lacks the nuanced review needed for quality and strategic alignment.
*   **Changing Metadata Automatically:** While recommendations can be provided, automatic changes to titles, meta descriptions, or other metadata without human oversight could lead to unintended consequences.
*   **Removing Content:** Any form of content removal, whether text or media, requires careful human consideration.
*   **Business Decisions Without Review:** Any strategic business decisions stemming from these recommendations must undergo thorough human review and approval.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

This recommendation system is a tool for ongoing content optimization, and its efficacy is subject to change over time. Regular review and potential retraining are essential to maintain its relevance and accuracy.

### When to Review or Retrain the System

The system should be reviewed or retrained under the following observed conditions:

*   **Significant Traffic Changes:** Any measured, unexpected increases or decreases in organic traffic for a large segment of content. This may indicate a shift in user behavior or search engine algorithms.
*   **Click-Through Rate (CTR) Changes:** Observed significant shifts in average CTR, either upwards or downwards, could suggest that title and meta description recommendations need reassessment.
*   **Ranking Drops:** A directional decline in observed search engine rankings for key content, signaling that the factors influencing visibility may have changed.
*   **Introduction of New Content:** When a substantial volume of new content is published, the system should be re-evaluated to ensure it effectively integrates and makes recommendations for these new assets.
*   **Seasonal Trends:** During periods of observed strong seasonal shifts in search demand, the model's performance should be monitored closely, and retraining may be considered to account for these patterns.
*   **Google Algorithm Updates:** Following confirmed major Google algorithm updates, a comprehensive review and potential retraining of the system are highly recommended to adapt to new ranking factors.
*   **Quarterly Reviews:** Regular, scheduled quarterly reviews are essential for general maintenance, performance assessment, and identification of any subtle, long-term shifts in data patterns that may warrant retraining.
